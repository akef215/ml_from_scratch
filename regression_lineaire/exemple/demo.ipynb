```python
import pandas as pd
import matplotlib.pyplot as plt
from regression_lineaire.src.linear_regression import LinearRegression as LinReg
```

# Loading Data

```python
# Load dataset
data = pd.read_csv('../../data/train.csv')

# Uncomment to see data description
#print(data.describe()[['LotArea', 'SalePrice']])

# Use only 'LotArea' feature for simplicity and 'SalePrice' as target
# Plot the data
X_train = data[['LotArea']]
Y_train = data['SalePrice']
plt.figure()
plt.scatter(X_train, Y_train, color='blue', alpha=0.5)  
plt.xlabel('Lot Area')
plt.ylabel('Sale Price')
plt.title('Lot Area vs Sale Price')
plt.show()
```

# Normal Equation Method:
  For further understanding of the mathematical background behind it consult the docs/normal_equation_theory.pdf file

```python
# Fit the model using the Normal Equation method
LinRegModel = LinReg()
LinRegModel.fit(X_train, Y_train)

# Predict the coefficients and bias 
Y_pred = LinRegModel.predict(X_train)
(coef, bias) = LinRegModel.get_params()
print("Coefficients: ", coef[0])
print("bias: ", bias)
```

```python
# plot results
LinRegModel.plot_regression_1D(X_train, Y_train, Y_pred, "Normal Equation")
```

# Gradient Descent Algorithm
  For further understanding of the mathematical background behind it consult the docs/gradient_descent_theory.pdf file

```python
# Fit the model using the Gradient Descent method
# Note: You might want to scale your features for better convergence
# Here we use the default values for learning rate and iterations
# learning_rate=0.01, n_iterations=1000
# We normalize the data by default
# You can disable it by setting normalize=False

LinRegModel.fit(X_train, Y_train, method='gradient')
# Predict the coefficients and the bias
Y_pred = LinRegModel.predict(X_train)
coef, bias = LinRegModel.get_params()
print("Coefficients: ", coef.iloc[0])
print("bias: ", bias)
```

```python
# Get the history of the cost function and plot it
# We call it the learing curve
cost_history = LinRegModel.get_history()

LinRegModel.plot_cost_history()
```

```python
# plot results
LinRegModel.plot_regression_1D(X_train, Y_train, Y_pred, 'Gradient Descent')
```

```python
# We can test with a different learning rate and number of iterations
LinRegModel.fit(X_train, Y_train, method='gradient', learning_rate=0.0001, max_iter=5000)

# Predict the coefficients and the bias
Y_pred = LinRegModel.predict(X_train)
coef, bias = LinRegModel.get_params()
print("Coefficients: ", coef.iloc[0])
print("bias: ", bias)
```

```python
# plot results
LinRegModel.plot_regression_1D(X_train, Y_train, Y_pred, 'Gradient Descent')
LinRegModel.plot_cost_history()
```

```python
# We test the case of normalise = False
# Note: You might need to adjust the learning rate to a lower value
# since the features are not normalized and can have very different scales

LinRegModel.fit(X_train, Y_train, method='gradient', normalise=False, learning_rate=1e-20, max_iter=10000)

# Predict the coefficients and the bias
Y_pred = LinRegModel.predict(X_train)
coef, bias = LinRegModel.get_params()
print("Coefficients: ", coef.iloc[0])
print("bias: ", bias)
```

```python
# plot results
LinRegModel.plot_regression_1D(X_train, Y_train, Y_pred, "Gradient Descent")
```

```python
# We test the case of normalise = False
# With default learning rate and iterations

LinRegModel.fit(X_train, Y_train, method='gradient', normalise=False)

# Predict the coefficients and the bias
coef, bias = LinRegModel.get_params()
print("Coefficients: ", coef.iloc[0])
print("bias: ", bias)
```
